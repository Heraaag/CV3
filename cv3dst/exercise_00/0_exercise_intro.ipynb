### Local install - pip or uv

Run the following commands in a terminal in the `cv3dst` folder. On Windows, activate the environment with `.venv\Scripts\activate` instead.

With **pip** (requires Python 3.12):
```bash
python3.12 -m venv .venv
source .venv/bin/activate
pip install -r requirements.txt jupyter
```

With **uv** (downloads Python 3.12 if necessary):
```bash
uv venv --python 3.12
source .venv/bin/activate
uv pip install -r requirements.txt jupyter
```

Then start Jupyter with `jupyter notebook` from the activated environment. Check out the README for more details, e.g. how to install a CPU-only PyTorch or one for a different CUDA version.
